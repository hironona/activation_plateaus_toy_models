```python
# Save checkpoints to zip and download

import os
import shutil
from google.colab import files

project_dir = "/content/activation_plateaus_geometry"
checkpoints_dir = os.path.join(project_dir, "checkpoints")

if not os.path.isdir(checkpoints_dir):
    raise FileNotFoundError(f"Folder not found: {checkpoints_dir}")

ckpt_type = "noise-0.075"

# Creates /content/checkpoints_all.zip containing the whole checkpoints folder
zip_path = shutil.make_archive(
    base_name=f"{project_dir}/{ckpt_type}",
    format="zip",
    root_dir=project_dir,
    base_dir=f"checkpoints/class_spiral/{ckpt_type}/",
)

print(f"Created: {zip_path}")
files.download(zip_path)
```
